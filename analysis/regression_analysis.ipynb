#Basic model: use citations to represent seniority (the default model)

```
import pandas as pd
import numpy as np
import statsmodels.formula.api as smf
df = pd.read_csv("the path to data")

text1='citation_pct'
cols = [
    "choose_to_pursue", "field3", "novelty_raw",
    "feasibility_raw", "probability_raw",
    "uses_ai_method_log",
    "idea_similarity", "email",text1
]

numeric_cols = [
    "choose_to_pursue",
    "novelty_raw",
    "feasibility_raw",
    "probability_raw",
    "uses_ai_method_log",
    "idea_similarity",
    text1
]

for c in numeric_cols:
    df[c] = pd.to_numeric(df[c], errors="coerce")

df_clean = df[cols].dropna().copy()

varying_vars = ["novelty_raw", "feasibility_raw", "probability_raw", "idea_similarity"]
for v in varying_vars:
    mean_v = df_clean.groupby("email")[v].transform("mean")
    df_clean[f"{v}_b"] = mean_v
    df_clean[f"{v}_w"] = df_clean[v] - mean_v

formula = """
choose_to_pursue ~
  novelty_raw_w + novelty_raw_b
+ feasibility_raw_w + feasibility_raw_b
+ probability_raw_w + probability_raw_b
+ idea_similarity_w + idea_similarity_b
+ citation_pct + uses_ai_method_log
+ C(field3, Treatment(reference='social'))
"""

ols_model = smf.ols(formula, data=df_clean)
ols_res = ols_model.fit(
    cov_type="cluster",
    cov_kwds={"groups": df_clean["email"]}
)

print(ols_res.summary())
```

#The ordinal regression version of the basic model

```
import pandas as pd
import numpy as np
from statsmodels.miscmodels.ordinal_model import OrderedModel
df = pd.read_csv("the path to data")

cols = [
    "choose_to_pursue", "field3", "novelty_raw",
    "feasibility_raw", "probability_raw",
    "citation_pct", "uses_ai_method_log",
    "idea_similarity", "email"
]

numeric_cols = [
    "choose_to_pursue",
    "novelty_raw",
    "feasibility_raw",
    "probability_raw",
    "citation_pct",
    "uses_ai_method_log",
    "idea_similarity"
]

for c in numeric_cols:
    df[c] = pd.to_numeric(df[c], errors="coerce")

df_clean = df[cols].dropna().copy()

# Mundlak decomposition: within / between
varying_vars = ["novelty_raw", "feasibility_raw", "probability_raw", "idea_similarity"]
for v in varying_vars:
    mean_v = df_clean.groupby("email")[v].transform("mean")
    df_clean[f"{v}_b"] = mean_v
    df_clean[f"{v}_w"] = df_clean[v] - mean_v

df_clean["choose_to_pursue"] = df_clean["choose_to_pursue"].astype(int)
ordered_cats = sorted(df_clean["choose_to_pursue"].unique())
df_clean["choose_to_pursue"] = pd.Categorical(
    df_clean["choose_to_pursue"],
    categories=ordered_cats,
    ordered=True
)

# ---------- exogenous variables ----------
xvars = [
    "novelty_raw_w", "novelty_raw_b",
    "feasibility_raw_w", "feasibility_raw_b",
    "probability_raw_w", "probability_raw_b",
    "idea_similarity_w", "idea_similarity_b",
    "citation_pct", "uses_ai_method_log"
]

X = df_clean[xvars].copy()

# field fixed effects via dummies
field_dummies = pd.get_dummies(
    df_clean["field3"],
    prefix="field3",
    drop_first=True,
    dtype=float
)

X = pd.concat([X, field_dummies], axis=1)
model = OrderedModel(
    endog=df_clean["choose_to_pursue"],
    exog=X,
    distr="logit"
)
res_cluster = model.fit(
    method="bfgs",
    disp=False,
    cov_type="cluster",
    cov_kwds={"groups": df_clean["email"]}
)

print(res_cluster.summary())
```

# Replication of GraphEval_GNN

```
"""
GraphEval-GNN for single-hypothesis novelty scoring
====================================================

Faithful adaptation of GraphEval-GNN (Feng et al., ICLR 2025,
"GraphEval: A Lightweight Graph-Based LLM Framework for Idea Evaluation")
to the setting where each idea is a *single* short AI-generated hypothesis.

What changes vs. the original paper
------------------------------------
* In the paper a long abstract is decomposed by a prompted LLM into many
  fine-grained "viewpoint" nodes. Our hypotheses are 1-3 sentences, so we
  treat each hypothesis as ONE atomic node (no viewpoint decomposition).
* Nodes are connected by top-5 cosine-similarity edges across hypotheses
  forming a hypothesis-similarity graph.
* A two-layer weighted GNN (Eq. 3 of the paper) learns node representations;
  since each idea is a single node, the within-idea MEAN/MAX pooling of Eq. 4
  collapses to the node embedding itself, so the head is just MLP(softmax)
  applied to the node embedding.
* The 1-9 human novelty rating is mapped into the four review-decision
  classes (reject / poster / oral / spotlight) by *within-field quantile*
  binning (25/50/75 percentiles, rounded), with cut-point shifting to keep
  all four classes non-empty.
* Training is transductive *within each field*: the whole field is one graph;
  training hypotheses carry binned labels, test hypotheses are unlabeled.
* The scalar novelty score collapses the class distribution via
      Score = P(spotlight) + P(oral) + P(poster)

Required input
--------------
A pandas DataFrame with columns:
    df['hypothesis']   : str   (the idea text)
    df['field3']       : any   (grouping key; quantiles are per-field)
    df['novelty_raw']  : 1-9   (ground-truth novelty rating)

Output
------
The same DataFrame with added columns:
    grapheval_score        : float novelty score in [0, 1]

Dependencies
------------
    pip install torch torch_geometric scikit-learn pandas numpy scipy
    pip install transformers          # only for the default BERT embedder

Note: the default embedder downloads a BERT checkpoint from Hugging Face on
first use (needs internet). If you already have embeddings, pass them via the
`embeddings=` argument and the download is skipped entirely.
"""

from __future__ import annotations

import warnings
from typing import Optional

import numpy as np
import pandas as pd
import torch
import torch.nn as nn
import torch.nn.functional as F
from sklearn.model_selection import StratifiedKFold
from torch_geometric.nn import MessagePassing

# Ordered class names: index 0..3 maps to increasing novelty.
CLASS_NAMES = ["reject", "poster", "oral", "spotlight"]


# --------------------------------------------------------------------------- #
# 1. Embedding                                                                #
# --------------------------------------------------------------------------- #
def embed_hypotheses(
    texts,
    model_name: str = "bert-base-uncased",
    device: Optional[str] = None,
    batch_size: int = 64,
    max_length: int = 128,
) -> np.ndarray:
    """Embed hypotheses with a BERT encoder using mean pooling over tokens.

    Faithful to the paper's "BERT-based encoder". You can swap `model_name`
    for any Hugging Face encoder (e.g. a sentence-transformers checkpoint).
    Returns an (n, hidden) float32 array.
    """
    from transformers import AutoModel, AutoTokenizer  # local import

    device = device or ("cuda" if torch.cuda.is_available() else "cpu")
    tok = AutoTokenizer.from_pretrained(model_name)
    model = AutoModel.from_pretrained(model_name).to(device).eval()

    texts = [str(t) for t in texts]
    out = []
    with torch.no_grad():
        for i in range(0, len(texts), batch_size):
            batch = texts[i : i + batch_size]
            enc = tok(
                batch,
                padding=True,
                truncation=True,
                max_length=max_length,
                return_tensors="pt",
            ).to(device)
            hidden = model(**enc).last_hidden_state          # (b, L, h)
            mask = enc["attention_mask"].unsqueeze(-1).float()
            pooled = (hidden * mask).sum(1) / mask.sum(1).clamp(min=1e-9)
            out.append(pooled.cpu().numpy())
    return np.concatenate(out, axis=0).astype(np.float32)


# --------------------------------------------------------------------------- #
# 2. Within-field quantile binning                                            #
# --------------------------------------------------------------------------- #
def bin_ratings_within_field(ratings) -> tuple[np.ndarray, tuple]:
    """Bin 1-9 novelty ratings into 4 ordered classes using within-field
    quantiles. Cut points c1<=c2<=c3 are the rounded 25/50/75 percentiles.

    Assignment:  r<=c1 -> reject(0); c1<r<=c2 -> poster(1);
                 c2<r<=c3 -> oral(2); r>c3 -> spotlight(3).

    Returns (labels[int array], (c1, c2, c3)).
    """
    r = np.asarray(ratings, dtype=float)
    n = len(r)
    if n == 0:
        return np.array([], dtype=int), (None, None, None)

    c1 = int(round(np.percentile(r, 25)))
    c2 = int(round(np.percentile(r, 50)))
    c3 = int(round(np.percentile(r, 75)))
    lo, hi = int(np.floor(r.min())), int(np.ceil(r.max()))

    def assign(a, b, c):
        lab = np.empty(n, dtype=int)
        lab[r <= a] = 0
        lab[(r > a) & (r <= b)] = 1
        lab[(r > b) & (r <= c)] = 2
        lab[r > c] = 3
        return lab

    def four_nonempty(a, b, c):
        return a < b < c and len(np.unique(assign(a, b, c))) == 4

    # Try the rounded percentiles, forced strictly increasing first.
    a, b, c = c1, max(c2, c1 + 1), max(max(c2, c1 + 1) + 1, c3)
    if four_nonempty(a, b, c):
        return assign(a, b, c), (a, b, c)

    # Otherwise search integer triples; maximize the smallest class, then
    # minimize size variance -- a principled generalization of "shift by a
    # single rating value" that guarantees four non-empty, balanced classes.
    best, best_key = None, None
    for x in range(lo, hi):
        for y in range(x + 1, hi):
            for z in range(y + 1, hi):
                lab = assign(x, y, z)
                if len(np.unique(lab)) == 4:
                    sizes = np.bincount(lab, minlength=4)
                    key = (sizes.min(), -float(sizes.var()))
                    if best_key is None or key > best_key:
                        best_key, best = key, (x, y, z)
    if best is not None:
        return assign(*best), best

    # Not enough distinct values for four ordered bins -> rank fallback.
    warnings.warn(
        "A field has too few distinct novelty ratings to form four "
        "quantile bins; falling back to even rank binning for it.",
        RuntimeWarning,
    )
    order = np.argsort(r, kind="stable")
    lab = np.empty(n, dtype=int)
    edges = np.linspace(0, n, 5).astype(int)
    for cls in range(4):
        lab[order[edges[cls] : edges[cls + 1]]] = cls
    return lab, (c1, c2, c3)


# --------------------------------------------------------------------------- #
# 3. Top-k cosine-similarity graph                                            #
# --------------------------------------------------------------------------- #
def build_topk_graph(emb: np.ndarray, k: int = 5):
    """Top-k cosine-similarity, undirected, weighted graph.

    Each node connects to its k most similar peers; edges are symmetrized and
    weighted by cosine similarity. Returns torch (edge_index[2,E], edge_weight[E]).
    """
    n = emb.shape[0]
    k = max(1, min(k, n - 1))
    norm = emb / (np.linalg.norm(emb, axis=1, keepdims=True) + 1e-12)
    sim = norm @ norm.T
    np.fill_diagonal(sim, -np.inf)                       # no self-loops
    nbr = np.argpartition(-sim, kth=k - 1, axis=1)[:, :k]

    edges: dict[tuple, float] = {}
    for i in range(n):
        for j in nbr[i]:
            j = int(j)
            a, b = (i, j) if i < j else (j, i)
            w = float(sim[i, j])
            edges[(a, b)] = max(edges.get((a, b), -np.inf), w)

    src, dst, wt = [], [], []
    for (a, b), w in edges.items():
        src += [a, b]
        dst += [b, a]
        wt += [w, w]
    edge_index = torch.tensor([src, dst], dtype=torch.long)
    edge_weight = torch.tensor(wt, dtype=torch.float32)
    return edge_index, edge_weight

# --------------------------------------------------------------------------- #
# 4. Weighted GraphConv (Eq. 3) and the GNN                                   #
# --------------------------------------------------------------------------- #
class WeightedGraphConv(MessagePassing):
    r"""h_v = U( CONCAT( MEAN_{q in N(v)} ReLU(w_vq * W h_q), h_v ) )."""

    def __init__(self, in_ch: int, out_ch: int):
        super().__init__(aggr="mean")
        self.lin_neighbor = nn.Linear(in_ch, out_ch, bias=False)   # W
        self.lin_update = nn.Linear(out_ch + in_ch, out_ch)        # U on concat

    def forward(self, x, edge_index, edge_weight):
        agg = self.propagate(edge_index, x=x, edge_weight=edge_weight)
        return self.lin_update(torch.cat([agg, x], dim=-1))

    def message(self, x_j, edge_weight):
        return F.relu(edge_weight.view(-1, 1) * self.lin_neighbor(x_j))

class GraphEvalGNN(nn.Module):
    """Two-layer weighted GNN + MLP classification head over 4 classes."""
    """In our experiments, we found that mapping to 1–9 classes was ineffective and unstable."""

    def __init__(self, in_dim: int, hidden: int = 64, n_classes: int = 4,
                 dropout: float = 0.5):
        super().__init__()
        self.conv1 = WeightedGraphConv(in_dim, hidden)
        self.conv2 = WeightedGraphConv(hidden, hidden)
        self.dropout = dropout
        # Single-node subgraph: MEAN/MAX pooling collapse to the node embedding.
        self.head = nn.Sequential(
            nn.Linear(hidden, hidden), nn.ReLU(),
            nn.Linear(hidden, n_classes),
        )

    def forward(self, x, edge_index, edge_weight):
        x = F.relu(self.conv1(x, edge_index, edge_weight))
        x = F.dropout(x, p=self.dropout, training=self.training)
        x = self.conv2(x, edge_index, edge_weight)
        return self.head(x)                              # logits (n, 4)


# --------------------------------------------------------------------------- #
# 5. Transductive train + predict on one field graph                          #
# --------------------------------------------------------------------------- #
def _train_predict(
    emb: np.ndarray,
    labels: np.ndarray,
    train_mask: np.ndarray,
    hidden: int = 64,
    k: int = 5,
    epochs: int = 300,
    lr: float = 1e-3,
    weight_decay: float = 5e-4,
    dropout: float = 0.5,
    device: Optional[str] = None,
    seed: int = 0,
) -> np.ndarray:
    """Build the full field graph, train on `train_mask`, return softmax
    probabilities for ALL nodes. Transductive: every node is in the
    graph; only train nodes contribute to the loss."""
    device = device or ("cuda" if torch.cuda.is_available() else "cpu")
    torch.manual_seed(seed)

    edge_index, edge_weight = build_topk_graph(emb, k=k)
    x = torch.tensor(emb, dtype=torch.float32, device=device)
    edge_index = edge_index.to(device)
    edge_weight = edge_weight.to(device)
    y = torch.tensor(labels, dtype=torch.long, device=device)
    tm = torch.tensor(train_mask, dtype=torch.bool, device=device)

    # Class weights guard against any residual imbalance within the train split.
    counts = np.bincount(labels[train_mask], minlength=4).astype(float)
    cls_w = torch.tensor(
        np.where(counts > 0, counts.sum() / (4.0 * np.maximum(counts, 1)), 0.0),
        dtype=torch.float32, device=device,
    )

    model = GraphEvalGNN(emb.shape[1], hidden=hidden, dropout=dropout).to(device)
    opt = torch.optim.Adam(model.parameters(), lr=lr, weight_decay=weight_decay)
    sched = torch.optim.lr_scheduler.LambdaLR(
        opt, lambda e: max(0.0, 1.0 - e / max(1, epochs))
    )

    model.train()
    for _ in range(epochs):
        opt.zero_grad()
        logits = model(x, edge_index, edge_weight)
        loss = F.cross_entropy(logits[tm], y[tm], weight=cls_w)
        loss.backward()
        opt.step()
        sched.step()

    model.eval()
    with torch.no_grad():
        probs = F.softmax(model(x, edge_index, edge_weight), dim=-1)
    return probs.cpu().numpy()


# --------------------------------------------------------------------------- #
# 6. End-to-end driver                                                        #
# --------------------------------------------------------------------------- #
def run_grapheval_gnn(
    df: pd.DataFrame,
    embeddings: Optional[np.ndarray] = None,
    model_name: str = "bert-base-uncased",
    k: int = 5,
    n_splits: int = 5,
    hidden: int = 64,
    epochs: int = 300,
    lr: float = 1e-3,
    weight_decay: float = 5e-4,
    dropout: float = 0.5,
    device: Optional[str] = None,
    seed: int = 0,
    text_col: str = "hypothesis",
    field_col: str = "field3",
    rating_col: str = "novelty_raw",
) -> pd.DataFrame:

    df = df.reset_index(drop=True).copy()

    if embeddings is None:
        embeddings = embed_hypotheses(
            df[text_col].tolist(), model_name=model_name, device=device
        )
    embeddings = np.asarray(embeddings, dtype=np.float32)
    assert len(embeddings) == len(df), "embeddings must align with df rows"

    df["grapheval_label"] = -1
    for cls in CLASS_NAMES:
        df[f"p_{cls}"] = np.nan

    for field, idx in df.groupby(field_col).groups.items():
        idx = np.asarray(list(idx))
        emb_f = embeddings[idx]
        ratings_f = df.loc[idx, rating_col].to_numpy()

        labels_f, cuts = bin_ratings_within_field(ratings_f)
        df.loc[idx, "grapheval_label"] = labels_f

        n = len(idx)
        if n < 4:
            warnings.warn(
                f"Field {field!r} has only {n} hypotheses; skipping (need >=4).",
                RuntimeWarning,
            )
            continue

        # Adapt fold count to the rarest class so stratification is valid.
        # StratifiedKFold needs >=folds samples per class; if the rarest class
        # has <2 members we cannot CV this field, so skip it with a warning
        # rather than crash.
        min_cls = int(np.bincount(labels_f, minlength=4).min())
        if min_cls < 2:
            warnings.warn(
                f"Field {field!r}: rarest class has <2 samples; skipping CV.",
                RuntimeWarning,
            )
            continue
        folds = min(n_splits, min_cls)
        skf = StratifiedKFold(n_splits=folds, shuffle=True, random_state=seed)

        probs_field = np.zeros((n, 4), dtype=np.float32)
        for tr, te in skf.split(np.arange(n), labels_f):
            train_mask = np.zeros(n, dtype=bool)
            train_mask[tr] = True
            probs = _train_predict(
                emb_f, labels_f, train_mask,
                hidden=hidden, k=k, epochs=epochs, lr=lr,
                weight_decay=weight_decay, dropout=dropout,
                device=device, seed=seed,
            )
            probs_field[te] = probs[te]            # keep only out-of-fold rows

        for c, cls in enumerate(CLASS_NAMES):
            df.loc[idx, f"p_{cls}"] = probs_field[:, c]

    df["grapheval_pred_class"] = df[[f"p_{c}" for c in CLASS_NAMES]].to_numpy().argmax(1)
    df["grapheval_score"] = df[["p_poster", "p_oral", "p_spotlight"]].sum(axis=1)
    return df

# --------------------------------------------------------------------------- #
# 7. main                                                                     #
# --------------------------------------------------------------------------- #
def main():
    import argparse

    p = argparse.ArgumentParser(description="GraphEval-GNN novelty scoring.")
    p.add_argument("--input", default="path to data",
                   help="Input CSV with hypothesis/field3/novelty_raw columns.")
    p.add_argument("--output", default=None,
                   help="Output CSV. Default: <input>_scored.csv")
    p.add_argument("--embeddings", default=None,
                   help="Optional .npy (n,dim) aligned with the CSV; skips BERT.")
    p.add_argument("--text-col", default="hypothesis")
    p.add_argument("--field-col", default="field3")
    p.add_argument("--rating-col", default="novelty_raw")
    p.add_argument("--model-name", default="bert-base-uncased")
    p.add_argument("--device", default=None, help="'cuda', 'cpu', or auto.")
    p.add_argument("--k", type=int, default=5)
    p.add_argument("--n-splits", type=int, default=5)
    p.add_argument("--hidden", type=int, default=64)
    p.add_argument("--epochs", type=int, default=300)
    p.add_argument("--lr", type=float, default=1e-3)
    p.add_argument("--weight-decay", type=float, default=5e-4)
    p.add_argument("--dropout", type=float, default=0.5)
    p.add_argument("--seed", type=int, default=0)
    args = p.parse_args()

    df = pd.read_csv(args.input)
    print(f"[info] loaded {len(df)} rows from {args.input}", flush=True)

    missing = [c for c in (args.text_col, args.field_col, args.rating_col)
               if c not in df.columns]
    if missing:
        raise SystemExit(f"[error] missing column(s) {missing}; "
                         f"available: {list(df.columns)}")

    embeddings = None
    if args.embeddings is not None:
        embeddings = np.load(args.embeddings)
        if len(embeddings) != len(df):
            raise SystemExit(f"[error] embeddings rows ({len(embeddings)}) "
                             f"!= df rows ({len(df)})")
        print(f"[info] using precomputed embeddings {embeddings.shape}", flush=True)

    out = run_grapheval_gnn(
        df, embeddings=embeddings, model_name=args.model_name,
        k=args.k, n_splits=args.n_splits, hidden=args.hidden, epochs=args.epochs,
        lr=args.lr, weight_decay=args.weight_decay, dropout=args.dropout,
        device=args.device, seed=args.seed,
        text_col=args.text_col, field_col=args.field_col, rating_col=args.rating_col,
    )

    out_path = args.output
    if out_path is None:
        import os
        base, ext = os.path.splitext(args.input)
        out_path = f"{base}_scored{ext or '.csv'}"
    out.to_csv(out_path, index=False)
    print(f"[info] wrote scored CSV -> {out_path}", flush=True)

    cols = [args.field_col, args.rating_col, "grapheval_label",
            "grapheval_pred_class", "grapheval_score"]
    print("\n[preview]")
    print(out[cols].head(10).to_string(index=False))

if __name__ == "__main__":
    main()
```
